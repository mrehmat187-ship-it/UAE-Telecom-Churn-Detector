```python
import pandas as pd



telecom_data = {
    'Customer_ID': tuple((501, 502, 503, 504, 505, 506)),
    'Plan': tuple(('Postpaid', 'Prepaid', 'Postpaid', 'Prepaid', 'Postpaid', 'Prepaid')),
    'Monthly_Bill': tuple((450, 80, 620, 60, 510, 95)),
    'Complaints': tuple((3, 0, 4, 1, 0, 3)),
    'Churned': tuple(('Yes', 'No', 'Yes', 'No', 'No', 'Yes'))
}

df = pd.DataFrame(telecom_data)

print("--- Customer Churn Count ---")
print(df['Churned'].value_counts())

high_risk = df[(df['Complaints'] >= 3) & (df['Monthly_Bill'] > 300)]
print("\n--- High-Value At-Risk Customer ---")
print(high_risk)
```
